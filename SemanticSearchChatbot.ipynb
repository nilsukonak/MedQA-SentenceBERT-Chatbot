```python
questions = df["Question"].tolist()
answers = df["Answer"].tolist()

print(questions[0])
print(answers[0])
```

```python
from sentence_transformers import SentenceTransformer

# basit ve hızlı model seçiyoruz
model = SentenceTransformer("all-MiniLM-L6-v2")

# tüm soruları vektöre çeviriyoruz
question_embeddings = model.encode(questions)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def cevap_bul(soru):
    soru_vec = model.encode([soru])


    benzerlikler = cosine_similarity(soru_vec, question_embeddings)[0]

    en_iyi_index = benzerlikler.argmax()

    return answers[en_iyi_index]
```

```python
print(cevap_bul("Can I take this medicine with food?"))
```

```python
#ARAYÜZ
import gradio as gr

def chatbot(soru):
    cevap = cevap_bul(soru)
    return cevap

arayuz = gr.Interface(
    fn=chatbot,
    inputs=gr.Textbox(lines=2, placeholder="Sorunuzu yazın..."),
    outputs="text",
    title="İlaç Soru cevap chatbot (Model-1)",
    description="Sentence-BERT tabanlı benzerlik modeli"
)

arayuz.launch()
```
